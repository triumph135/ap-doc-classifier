```python
# Packages
from transformers import pipeline
import pandas as pd
```

```python
# Instantiate zero shot image classification model
checkpoint = "openai/clip-vit-large-patch14"
detector = pipeline(model=checkpoint, task="zero-shot-image-classification")
```

```python
# Apply to my document
predictions = detector("test_bol2.png", candidate_labels=["Bill of Lading Invoice", "Bill of Lading", "Purchase Order Receipt", "Purchase Order Invoice"])
predictions
```
